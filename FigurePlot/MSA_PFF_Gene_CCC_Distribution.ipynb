```python
import os
from pathlib import Path

import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns


# -------------------------
# PDF compatibility
# -------------------------
# Prevents some PDF readers from reporting:
# "Insufficient data for an image"
mpl.rcParams["pdf.compression"] = 0


# -------------------------
# Project paths
# -------------------------
# The Jupyter Notebook working directory should be:
# Code/Figures Python
project_dir = Path.cwd()

data_dir = project_dir / "Data"
results_dir = project_dir / "Results"


# -------------------------
# Input file
# -------------------------
file_xlsx = (
    data_dir
    / "MSA_PFF_Gene_CCC_Distribution.xlsx"
)


# ============================================================
# MSA gene-model CCC distribution
# ============================================================
df = pd.read_excel(
    file_xlsx,
    sheet_name="MSA",
    header=0
)

# Create a new figure explicitly.
fig, ax = plt.subplots()

sns.histplot(
    data=df,
    x="CCC",
    stat="proportion",
    kde=True,
    bins=50,
    color=sns.color_palette("dark:#5A9_r")[0],
    ax=ax
)

# SNCA result
ax.axvline(
    0.516684383861818,
    color=[0.706, 0.016, 0.150],
    linestyle="--"
)

# GBA result
ax.axvline(
    0.473509503,
    color=[0.229, 0.298, 0.754],
    linestyle="--"
)

# MSA best gene: Kctd4
ax.axvline(
    0.643216732541456,
    color=[0.91, 0.35, 0.02],
    linestyle="--"
)

# PFF best gene: Dok5 evaluated in the MSA model
ax.axvline(
    0.417493669253098,
    color=[0.0, 0.47, 0.44],
    linestyle="--"
)

# Global MSA model
ax.axvline(
    0.395578415583938,
    color=[0.85, 0.53, 0.10],
    linestyle="-"
)

ax.set_xlabel("Model performance (CCC)")
ax.set_ylabel("Proportion")
ax.set_title("MSA gene-model CCC distribution")

# Optional fixed x-axis:
# ax.set_xlim(0, 1)

plt.tight_layout()
plt.show()


# -------------------------
# Save MSA figure
# -------------------------
FolderName = (
    results_dir
    / "MSA_PFF_Gene_CCC_Distribution"
    / "MSA"
)

FolderName.mkdir(
    parents=True,
    exist_ok=True
)

NameForSave = "MSA_Gene"

SaveFileTif = (
    FolderName
    / f"{NameForSave}_Distribution.tif"
)

SaveFilePdf = (
    FolderName
    / f"{NameForSave}_Distribution.pdf"
)

fig.savefig(
    SaveFilePdf,
    format="pdf"
)

fig.savefig(
    SaveFileTif,
    format="tif",
    dpi=300,
    facecolor="white",
    transparent=False
)

plt.close(fig)

print("Saved:")
print(SaveFilePdf)
print(SaveFileTif)


# ============================================================
# PFF gene-model CCC distribution
# ============================================================
df = pd.read_excel(
    file_xlsx,
    sheet_name="PFF",
    header=0
)

# Create a separate figure for PFF.
fig, ax = plt.subplots()

sns.histplot(
    data=df,
    x="CCC",
    stat="proportion",
    kde=True,
    bins=50,
    color=sns.color_palette("dark:#5A9_r")[0],
    ax=ax
)

# SNCA result
ax.axvline(
    0.691705675304722,
    color=[0.706, 0.016, 0.150],
    linestyle="--"
)

# GBA result
ax.axvline(
    0.677158899335937,
    color=[0.229, 0.298, 0.754],
    linestyle="--"
)

# MSA best gene: Kctd4 evaluated in the PFF model
ax.axvline(
    0.637441016984004,
    color=[0.91, 0.35, 0.02],
    linestyle="--"
)

# PFF best gene: Dok5
ax.axvline(
    0.73932787776396,
    color=[0.0, 0.47, 0.44],
    linestyle="--"
)

# Global PFF model
ax.axvline(
    0.647501625247243,
    color=[0.85, 0.53, 0.10],
    linestyle="-"
)

ax.set_xlabel("Model performance (CCC)")
ax.set_ylabel("Proportion")
ax.set_title("PFF gene-model CCC distribution")

# Optional fixed x-axis:
# ax.set_xlim(0, 1)

plt.tight_layout()
plt.show()


# -------------------------
# Save PFF figure
# -------------------------
FolderName = (
    results_dir
    / "MSA_PFF_Gene_CCC_Distribution"
    / "PFF"
)

FolderName.mkdir(
    parents=True,
    exist_ok=True
)

NameForSave = "PFF_Gene"

SaveFileTif = (
    FolderName
    / f"{NameForSave}_Distribution.tif"
)

SaveFilePdf = (
    FolderName
    / f"{NameForSave}_Distribution.pdf"
)

fig.savefig(
    SaveFilePdf,
    format="pdf"
)

fig.savefig(
    SaveFileTif,
    format="tif",
    dpi=300,
    facecolor="white",
    transparent=False
)

plt.close(fig)

print("Saved:")
print(SaveFilePdf)
print(SaveFileTif)
```
